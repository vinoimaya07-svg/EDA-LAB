```python
import pandas as pd
import folium
from folium.plugins import MarkerCluster
import plotly.express as px
```

```python
# Sample dataset
city_df = pd.DataFrame({
    'city': [
        'Chennai', 'Mumbai', 'Delhi',
        'Bengaluru', 'Kolkata', 'Hyderabad'
    ],
    'lat': [
        13.0827, 19.0760, 28.7041,
        12.9716, 22.5726, 17.3850
    ],
    'lon': [
        80.2707, 72.8777, 77.1025,
        77.5946, 88.3639, 78.4867
    ],
    'sales_lakhs': [
        120, 210, 195, 175, 90, 130
    ]
})
```

```python
# Step 1: Base interactive map

m = folium.Map(
    location=[22.0, 79.0],
    zoom_start=5,
    tiles='CartoDB positron'
)

```

```python
# Step 2: Marker cluster

marker_cluster = MarkerCluster().add_to(m)

for _, row in city_df.iterrows():

    folium.CircleMarker(
        location=[row['lat'], row['lon']],
        radius=8 + row['sales_lakhs'] / 20,
        color='crimson',
        fill=True,
        fill_opacity=0.7,
        tooltip=f"{row['city']}: Hover for details",
        popup=folium.Popup(
            f"<b>{row['city']}</b><br>"
            f"Sales: {row['sales_lakhs']} lakhs",
            max_width=200
        )
    ).add_to(marker_cluster)
```

```python
# Step 3: Layer control

folium.TileLayer(
    'OpenStreetMap'
).add_to(m)

folium.LayerControl().add_to(m)

# Step 4: Save map

m.save('india_sales_map.html')
```

```python
# ---------- Plotly Alternative ----------

fig = px.scatter_mapbox(
    city_df,
    lat='lat',
    lon='lon',
    size='sales_lakhs',
    color='sales_lakhs',
    hover_name='city',
    hover_data={
        'sales_lakhs': True,
        'lat': False,
        'lon': False
    },
    color_continuous_scale='YlOrRd',
    zoom=3.5,
    mapbox_style='carto-positron',
    title='City-wise Sales'
)

fig.update_layout(
    margin={
        'r': 0,
        't': 40,
        'l': 0,
        'b': 0
    }
)

fig.show()
```
